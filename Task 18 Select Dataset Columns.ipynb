# Task 18: Select Dataset Columns
**Objective:** Learn how to access specific parts of a DataFrame using Pandas `loc` and `iloc`.

**Deliverables:**
- Examples using `loc` (label-based selection)
- Examples using `iloc` (integer position-based selection)
- Output showing selected records from the Iris Dataset

```python
import pandas as pd

# 1. Load the Iris dataset
df = pd.read_csv('iris_dataset.csv')

print("--- Original DataFrame (First 3 rows) ---")
print(df.head(3))
print("\n")

# 2. Examples using loc (Label-based indexing)
# Selecting rows with index labels 0 to 4, and exact column names
print("--- Using loc: First 5 rows of sepal length and species ---")
loc_selection = df.loc[0:4, ['sepal length (cm)', 'species']]
print(loc_selection)
print("\n")

# 3. Examples using iloc (Integer position-based indexing)
# Selecting rows at integer positions 10 to 14, and columns at positions 0 and 4
print("--- Using iloc: Rows 10-14 for the 1st and 5th columns ---")
iloc_selection = df.iloc[10:15, [0, 4]]
print(iloc_selection)
```

### Interview Questions

**1. What is the difference between loc and iloc?**
*   `loc` is label-based data selecting method which means that we have to pass the name of the row or column which we want to select.
*   `iloc` is a purely integer-location based indexing for selection by position, meaning you have to pass integer indices (0-based) for the rows and columns you want to select.

**2. How do you select a single column in Pandas?**
You can select a single column by using bracket notation with the column name as a string (e.g., `df['column_name']`) or by using dot notation if the column name has no spaces (e.g., `df.column_name`).

**3. How do you select multiple columns?**
You can select multiple columns by passing a list of column names inside the bracket notation, like this: `df[['column_name_1', 'column_name_2']]`.
